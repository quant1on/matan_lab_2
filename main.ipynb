```python
true_val = np.exp(b) - np.exp(a)
```

```python
fig, ax = plt.subplots(1, 1, figsize=(8, 6))
ns = [2**i for i in range(9)]

ax.plot(ns, np.log(np.array([abs(rectangle_method(f, 'L', congruent_part(a, b, nn)) - true_val) for nn in ns])), color='purple', marker='o', label='Метод прямоугольников (левые)')
ax.plot(ns, np.log(np.array([abs(rectangle_method(f, 'M', congruent_part(a, b, nn)) - true_val) for nn in ns])), color='orange', marker='o', label='Метод прямоугольников (середина)')
ax.plot(ns, np.log(np.array([abs(trapezoid_method(f, congruent_part(a, b, nn)) - true_val) for nn in ns])), color='blue', marker='o', label='Метод трапеций')
ax.plot(ns, np.log(np.array([abs(simpson_method(f, congruent_part(a, b, nn)) - true_val) for nn in ns])), color='green', marker='o', label='Метод Симпсона')

ax.set_xlabel('n')
ax.set_label('MAE (log)')
ax.set_title('MAE (log) vs n')

ax.grid()
ax.legend()

plt.show()
```
